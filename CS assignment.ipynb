26. WAP to enter a num to check palidrome or not

```python
num = int(input("Enter a number = "))
original = num
reverse = 0

while num > 0:
    digit = num % 10        # last digit nikala kyuki modulus reminder deta h 
    reverse = reverse * 10 + digit   # reverse mein joda
    num = num // 10         # last digit hataya decimal ke baad wala  

if reverse == original:
    print("It is Palindrome")
else:
    print("It is not Palindrome")
```

<!-- Armstrong number kya hota hai?

Ek number Armstrong hota hai agar uske har digit ko uske total digits ki power se raise karke add karo, to sum wapas original number ke barabar aa jaaye.

Example samjho — 153:

153 mein 3 digits hain (1, 5, 3)
Har digit ko power 3 se raise karo (kyunki 3 digits hain):
1³ = 1
5³ = 125
3³ = 27
Sum = 1 + 125 + 27 = 153
Sum original number (153) ke barabar hai → Armstrong number hai!

Non-example — 123:

1³ + 2³ + 3³ = 1 + 8 + 27 = 36 -->

```python
n = int(input("Enter a Number = "))
length = 0
temp = n
while n != 0:
    length += 1
    n = n//10
print(f"Length Of {temp} is {length}")


sum  = 0 
n = temp 
while n != 0:
    r = n % 10
    sum = sum + (r ** length)
    n = n // 10
if sum == temp:
    print("Armstrong")
else:
    print("Not Armstrong")
```

33  WAP to find Even and Odd digits in num

```python
num = int(input("Enter a Number = "))
even_count = 0
odd_count = 0

while num > 0:
    digit = num % 10
    if digit % 2 == 0:
        even_count += 1
    else:
        odd_count += 1
    num = num // 10        

print("Even digits count:", even_count)
print("Odd digits count:", odd_count)
```
