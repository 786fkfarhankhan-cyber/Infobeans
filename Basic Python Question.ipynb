Q6. Find the area of a triangle, sides of which are 10 cm and 9 cm and the perimeter 36 cm.

```python
side1 = 10
side2 = 9
perimeter = 36
# perimeter ka matlab hota hai teeno sides ka total hota h (perimeter = side1 + side2 + side3)

side3 = perimeter - side1 - side2
# Normally triangle ka area hota hai - area = 1/2 × base × height
# Lekin yahan height diya hi nahi hai! Sirf teeno sides pata hain. Isi situation ke liye Heron's formula banaya gaya tha — jab teeno sides pata ho, height ki zaroorat hi nahi padti.

s = perimeter / 2

# area = √[ s × (s - side1) × (s - side2) × (s - side3) ]

area = (s * (s - side1) * (s - side2) * (s - side3)) ** 0.5

print("Third side =", side3, "cm")
print("Area =", area, "cm²")
```

Q12. Find the area of a right angled triangle whose hypotenuse is 13 cm and one of its sides containing the right angle is 12 cm. Find the length of the other side.

```python
# Given values
hypotenuse = 13
side1 = 12

# Step 1: Hypotenuse ko square karo (Pythagoras theorem ka pehla part)
hyp_square = hypotenuse ** 2

# Step 2: Side1 ko square karo
side1_square = side1 ** 2

# Step 3: Dono squares ka difference nikalo
# (Pythagoras: hypotenuse² = side1² + side2² → side2² = hypotenuse² - side1²)
difference = hyp_square - side1_square

# Step 4: Square root lagao taaki side2 (other side) mil jaye
side2 = difference ** 0.5

# Step 5: Area nikalo (right-angled triangle: area = 1/2 * base * height)
# Yahan side1 aur side2 ek dusre ke perpendicular hain, isliye ye formula chalega
area = 0.5 * side1 * side2

# Output
print("Other side =", side2, "cm")
print("Area =", area, "cm²")
```

Q14.The length of one of the diagonals of a field in the form of a quadrilateral is 46 m. The perpendicular distance of the other two vertices from the diagonal are 13 m and 10 m, find the area of the field.

```python
# Given values
diagonal = 46      # quadrilateral ki diagonal length
distance1 = 13      # pehle 1 triangle perpendicular distance
distance2 = 10      # dusre 2 triangle  perpendicular distance

# Step 1: Diagonal do triangles banata hai, dono ka area alag nikalo
area_triangle1 = 0.5 * diagonal * distance1
area_triangle2 = 0.5 * diagonal * distance2

# Step 2: Dono triangles ka area jodo -> yahi quadrilateral ka total area hai
area = area_triangle1 + area_triangle2

# Output
print("Area of triangle 1 =", area_triangle1, "m²")
print("Area of triangle 2 =", area_triangle2, "m²")
print("Total Area =", area, "m²")
```

Q33. A rectangular garden has dimensions of 30 m by 20 m and is divided into 4 parts by two pathways that run perpendicular from its sides. One pathway has a width of 3 m and the other, 4 m. Calculate the total usable area of the garden.

```python
length = 30
breadth = 20

path1_width = 3
path2_width = 4

garden_area = length * breadth

path1_area = length * path1_width
path2_area = breadth * path2_width

overlap = path1_width * path2_width

usable_area = garden_area - path1_area - path2_area + overlap

print("Usable area =", usable_area, "m²")
```

Q37. The surface of the cylinder is 149 cm². The cylinder height is 6 cm. What is the diameter of this cylinder? 
Assuming surface = total surface area:

```python
surface_area = 149
height = 6
pi = 3.14

a = 2 * pi
b = 2 * pi * height
c = -surface_area

radius = (-b + (b ** 2 - 4 * a * c) ** 0.5) / (2 * a)
diameter = 2 * radius

print("Diameter =", diameter, "cm")
```
